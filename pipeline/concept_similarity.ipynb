```python
import os
import pandas as pd
concept_files = []
genres = ['Electronic', 'Experimental', 'Folk', 'Hip-Hop', 'Instrumental', 'International', 'Pop', 'Rock']
base_dir = '/mnt/f/Code/Music_Dataset/sixth_concepts/'
concept_dirs = []
for g in genres:
    concept_dirs.append(base_dir + '/concepts_' + g + '/cavs/')

concept_vectors = {}
for d in concept_dirs:
    for root, dirs, files in os.walk(d):
        for file in files:
            if '_concept' in file:
                data = pd.read_pickle(root + file)
                concept_name = data['concepts'][0]
                # print('Cavs data is: ', len(data['cavs']), len(data['cavs'][0]), data['cavs'])
                if concept_name in concept_vectors:
                    concept_vectors[concept_name].append(data['cavs'][0])
                else:
                    concept_vectors[concept_name] = [data['cavs'][0]]
# print(concept_vectors)
```

```python
import numpy as np
concept_means = {}
for k, v in concept_vectors.items():
    concept_means[k] = np.mean(concept_vectors[k], axis=0)
```

```python
import pickle
filename = '/mnt/f/Code/Thesis/why-pop/data/all_concept_cav_means.pkl'
with open(filename, 'wb') as file:
    pickle.dump(concept_means, file)
```

```python
from sklearn.cluster import KMeans
data = np.array(list(concept_means.values()))
kmeans = KMeans(n_clusters=10).fit(data)
kmeans.labels_
```

```python
print('Concepts dict: ', concept_means)
```

```python
concept_clusters = list(zip(list(concept_means.keys()), kmeans.labels_))
print('Raw: ', concept_clusters)
```

```python
from sklearn.manifold import TSNE
centers = kmeans.cluster_centers_
# print(centers, centers.shape, type(centers))
clusters = np.array([i for i in range(10)], dtype=np.float64)
tsne = TSNE(n_components=2, init='random')
z = tsne.fit_transform(centers)
```

```python
z[0][0]
```

```python
# concept_clusters = list(zip(list(concept_means.keys()), kmeans.labels_))
concept_clusters_2 = []
for i, k in enumerate(concept_means.keys()):
    concept_clusters_2.append((k, kmeans.labels_[i], z[kmeans.labels_[i]][0], z[kmeans.labels_[i]][1]))
# print(z)
print(concept_clusters_2)
```

```python
import csv
filename = '/mnt/f/Code/Thesis/why-pop/src/data/all_concept_clusters.csv'
with open(filename, 'w') as file:
    csv_out = csv.writer(file)
    csv_out.writerow(['concept_name', 'concept_cluster', 'cluster_center_x', 'cluster_center_y'])
    for row in concept_clusters_2:
        csv_out.writerow(row)
```
